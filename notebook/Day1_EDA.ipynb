```python
monthly_sales_original = (
    df.groupby("YearMonth")["Sales"]
      .sum()
      .reset_index()
)

monthly_sales_original["YearMonth"] = monthly_sales_original["YearMonth"].astype(str)
```

```python
prophet_df = monthly_sales_original.rename(
    columns={
        "YearMonth": "ds",
        "Sales": "y"
    }
)

prophet_df["ds"] = pd.to_datetime(prophet_df["ds"])
```

```python
prophet_df.head()
```

```python
model = Prophet()
```

```python
model.fit(prophet_df)
```

```python
future = model.make_future_dataframe(periods=12, freq="M")
```

```python
future.tail()
```

```python
forecast = model.predict(future)
```

```python
forecast[["ds", "yhat", "yhat_lower", "yhat_upper"]].tail().round(2)
```

```python
fig = model.plot(forecast)
```

```python
fig2 = model.plot_components(forecast)
```

```python
forecast_eval = forecast[["ds", "yhat"]]
```

```python
evaluation = prophet_df.merge(forecast_eval, on="ds").round(0)
```

```python
evaluation.head()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from math import sqrt

mae = mean_absolute_error(evaluation["y"], evaluation["yhat"])

rmse = sqrt(
    mean_squared_error(evaluation["y"], evaluation["yhat"])
)

r2 = r2_score(evaluation["y"], evaluation["yhat"])

print("MAE :", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("R²  :", round(r2, 4))
```

```python
# ==========================================
# ANOMALY DETECTION
# ==========================================
```

```python
from sklearn.ensemble import IsolationForest
```

```python
features = [
    "Sales",
    "Profit",
    "Discount",
    "Quantity",
    "Shipping.Cost"
]

X = df[features]
```

```python
iso = IsolationForest(
    contamination=0.02,
    random_state=42
)

df["Anomaly"] = iso.fit_predict(X)
```

```python
anomaly_count = (df["Anomaly"] == -1).sum()
normal_count = (df["Anomaly"] == 1).sum()

print("="*60)
print("ANOMALY REPORT")
print("="*60)

print(f"Normal Orders : {normal_count}")
print(f"Anomalies     : {anomaly_count}")
print(f"Percentage    : {anomaly_count/len(df)*100:.2f}%")
```

```python
anomalies = df[df["Anomaly"] == -1]

display(anomalies.head(10))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(
    df["Sales"],
    df["Profit"],
    c=df["Anomaly"],
    cmap="coolwarm",
    alpha=0.6
)

plt.xlabel("Sales")
plt.ylabel("Profit")
plt.title("Anomaly Detection")

plt.show()
```

```python
def ask_decisioniq(question):

    prompt = f"""
You are DecisionIQ.

You are an AI Business Intelligence Assistant.

Answer ONLY using the provided business information.

Business Information:

{business_context}

User Question:

{question}

Rules:

• Give concise but insightful answers.
• Use bullet points where appropriate.
• Do not invent statistics.
• If information is unavailable, clearly state that.
• End with one practical recommendation when possible.
"""

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content
```
